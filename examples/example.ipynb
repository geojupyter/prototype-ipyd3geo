```python
def quake_color(properties):  # noqa: D103, ANN001, ANN201
    # Sequential yellow -> dark red ramp over the magnitude range (5.1-8.1).
    mag = properties["mag"]
    if mag >= 7:  # noqa: PLR2004
        return "#800026"
    if mag >= 6.5:  # noqa: PLR2004
        return "#bd0026"
    if mag >= 6:  # noqa: PLR2004
        return "#e31a1c"
    if mag >= 5.5:  # noqa: PLR2004
        return "#fd8d3c"
    return "#fed976"


def quake_radius(properties):  # noqa: D103, ANN001, ANN201
    return 2 * properties["mag"] - 7.6


m6 = Map(
    basemap=True,
    graticule=True,
    dynamic=False,
)
m6.add_from_jgis(
    "big_quakes.jGIS",
    style=VectorStyle(
        fill_color=quake_color,
        fill_opacity=0.8,
        stroke_color="#ffffff",
        weight=0.75,
        stroke_opacity=0.9,
        point_radius=quake_radius,
    ),
)
m6
```

```python
import geopandas as gpd

bermuda = gpd.GeoDataFrame(
    {"name": ["Bermuda Triangle"]},
    geometry=gpd.GeoSeries.from_wkt(
        ["POLYGON ((-80.19 25.76, -64.75 32.30, -66.11 18.47, -80.19 25.76))"],
        crs=4326,
    ),
)

m7 = Map(
    projection=Projection("geoInterruptedHomolosine", rotate=(70, -25, 0)),
    basemap=True,
    graticule=True,
    dynamic=False,
)
m7.add(
    GeoData(
        bermuda,
        name="bermuda",
        style=VectorStyle(
            stroke_color="#1f4e79",
            fill_color="#4a90d9",
            weight=1.5,
            fill_opacity=0.5,
        ),
    )
)
m7
```
